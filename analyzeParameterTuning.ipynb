# Analysis of variance based on **Seed** parameter

Given the dataframe enriched with **Instance** column, we now perform an analysis of variance on **Seed** parameter for several series (i.e., **ObjBest**, **%Gap (ObjBest)**, **Runtime**, **%Gap (Runtime)**, **RuntimeBest**, **%Gap (RuntimeBest)**), some of which must previously defined. The analysis is performed separately for each **Instance**.

```python
import os
import matplotlib.pyplot as plt
import seaborn as sns

# Set Seaborn theme
sns.set_theme(style='whitegrid', font_scale=1.1)

# Create new column '%Gap (ObjBest)' and add it to the DataFrame (i.e., by computing '%Gap (ObjBest)' per instance)
df['%Gap (ObjBest)'] = df.groupby('Instance')['ObjBest'].transform(lambda x: (x - x.max()) / x)

# Create new column '%Gap (Runtime)' and add it to the DataFrame (i.e., by computing '%Gap (Runtime)' per instance)
df['%Gap (Runtime)'] = df.groupby('Instance')['Runtime'].transform(lambda x: (x - x.min()) / x)

# Create new column '%Gap (RuntimeBest)' and add it to the DataFrame (i.e., by computing '%Gap (RuntimeBest)' per instance)
df['%Gap (RuntimeBest)'] = df.groupby('Instance')['RuntimeBest'].transform(lambda x: (x - x.min()) / x)

# Set up the list of columns for the analysis
cols = ['ObjBest', '%Gap (ObjBest)', 'Runtime', '%Gap (Runtime)', 'RuntimeBest', '%Gap (RuntimeBest)']

# Get the list of unique instances
instances = df['Instance'].unique()

# Create one figure per instance
for instance in instances:
    subset_df = df[df['Instance'] == instance]

    # Compute summary statistics grouped by 'Seed'
    summary_stats = subset_df.groupby('Seed')[cols].agg(['mean', 'min', 'max', 'std'])

    # Round to two decimal digits for readability
    summary_stats = summary_stats.round(4)
    
    # --- DEBUG ONLY ---
    # Print a summary of the main statistics
    # print(summary_stats)

    # Save the summary of the main statistics to a file (after defining the file path)
    outfile_path = os.path.expanduser("~/Development/Smart-Mater-Installation-Scheduling-Project/Smart-Mater-Installation-Scheduling-Project/output/Intel(R) Xeon(R) CPU E3-1245 v5 @ 3.50GHz/results_grasp/parameter_tuning/results_grasp_" + instance + "_summary_stats_seed.csv")
    summary_stats.to_csv(outfile_path)

    # Show the boxplot
    fig, axes = plt.subplots(3, 2, figsize=(15, 13))
    axes = axes.flatten()

    for ax, col in zip(axes, cols):
        sns.boxplot(
            data=subset_df, x='Seed', y=col, hue='Seed', palette='pastel', ax=ax, legend=False,
            width=0.6, fliersize=4, linewidth=1.2, boxprops=dict(alpha=0.8), 
        )
        ax.set_title(f"{col}", fontsize=12)
        ax.set_xlabel("Seed")
        ax.set_ylabel("")
        ax.grid(True, linestyle="--", alpha=0.5)

    fig.suptitle(f"Analysis of variance based on \'Seed\' parameter for \'{instance}\'", fontsize=16, fontweight="bold")
    plt.tight_layout(rect=[0, 0.03, 1, 0.95])
    plt.show()
    
    # Save the figure
    outfile_path = os.path.expanduser("~/Development/Smart-Mater-Installation-Scheduling-Project/Smart-Mater-Installation-Scheduling-Project/output/Intel(R) Xeon(R) CPU E3-1245 v5 @ 3.50GHz/results_grasp/parameter_tuning/results_grasp_" + instance + "_seed_variance.png")
    fig.savefig(outfile_path, dpi=100)
```
