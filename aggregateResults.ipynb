# Use **groupby** to create **mean_df_ins_1** write it to a CSV file.

```python
group_cols = ['Instance', 'Alpha', 'Beta', 'Chi', 'Delta', 'Epsilon']
```

```python
exclude_cols = ['MaxIter', 'Seed', 'TimeLimit']
```

```python
mean_cols = ['ObjBest', 'Runtime', 'NumIterations', 'NumIterations (to find the best solution)', 'RuntimeBest', 'NumIterationsFirstMethod', 'CumulativeRuntimeFirstMethod', 'NumIterationsSecondMethod', 'CumulativeRuntimeSecondMethod', 'NumIterationsThirdMethod', 'CumulativeRuntimeThirdMethod', 'NumIterationsFourthMethod', 'CumulativeRuntimeFourthMethod']
```

```python
print(mean_df_ins_1)
```
